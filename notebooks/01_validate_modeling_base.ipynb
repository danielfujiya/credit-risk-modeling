```sql
-- 1. Checar se a chave é única na tabela de target
SELECT
  COUNT(*) AS total_rows,
  COUNT(DISTINCT Customer_ID, date_position) AS distinct_keys
FROM workspace.credit_risk.gold_credit_modeling_base_target;
```

```sql
-- 2. Checar se a chave é única na tabela de features
SELECT
  COUNT(*) AS total_rows,
  COUNT(DISTINCT Customer_ID, date_position) AS distinct_keys
FROM workspace.credit_risk.gold_credit_modeling_base_features;
```

```sql
-- 3. Confirmar os valores existentes no alvo
SELECT
  flag_default,
  COUNT(*) AS quantity,
  ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 2) AS percentage
FROM workspace.credit_risk.gold_credit_modeling_base_target
GROUP BY flag_default
ORDER BY flag_default;
```

```sql
-- 4. Encontrar observações com target, mas sem features correspondentes
SELECT
  t.Customer_ID,
  t.date_position
FROM workspace.credit_risk.gold_credit_modeling_base_target AS t
LEFT ANTI JOIN workspace.credit_risk.gold_credit_modeling_base_features AS f
  ON t.Customer_ID = f.Customer_ID
 AND t.date_position = f.date_position
LIMIT 20;
```
