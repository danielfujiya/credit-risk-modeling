```sql
CREATE OR REPLACE TABLE workspace.credit_risk.gold_credit_modeling_base
USING DELTA
AS
SELECT *
FROM read_files(
  '/Volumes/workspace/credit_risk/source_files/train_credit.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);
```

```sql
create or replace table workspace.credit_risk.gold_credit_modeling_base_target as

select
Customer_ID
,case `month`
when 'January' then date('2025-01-01')
when 'February' then date('2025-02-01')
when 'March' then date('2025-03-01')
when 'April' then date('2025-04-01')
when 'May' then date('2025-05-01')
when 'June' then date('2025-06-01')
when 'July' then date('2025-07-01')
when 'August' then date('2025-08-01')
end as date_position
,if(Credit_Score = 'Poor',1,0) as flag_default
from workspace.credit_risk.gold_credit_modeling_base
```

```sql
create or replace table workspace.credit_risk.gold_credit_modeling_base_features as

select
Customer_ID
,case `month`
when 'January' then date('2025-01-01')
when 'February' then date('2025-02-01')
when 'March' then date('2025-03-01')
when 'April' then date('2025-04-01')
when 'May' then date('2025-05-01')
when 'June' then date('2025-06-01')
when 'July' then date('2025-07-01')
when 'August' then date('2025-08-01')
end as date_position
,Occupation
,try_cast(Annual_Income as float) Annual_Income
,try_cast(Monthly_Inhand_Salary as float) Monthly_Inhand_Salary
,try_cast(Num_Bank_Accounts as int) Num_Bank_Accounts
,try_cast(Num_Credit_Card as int) Num_Credit_Card
,try_cast(Interest_Rate as int) Interest_Rate
,try_cast(Num_of_Loan as int) Num_of_Loan
,Type_of_Loan
,Delay_from_due_date
,try_cast(Changed_Credit_Limit as float) Changed_Credit_Limit
,try_cast(Num_Credit_Inquiries as int) Num_Credit_Inquiries
,Credit_Mix
,try_cast(Outstanding_Debt as float) Outstanding_Debt
,try_cast(Credit_Utilization_Ratio as float) Credit_Utilization_Ratio
,Payment_of_Min_Amount
,try_cast(Total_EMI_per_month as float) Total_EMI_per_month
,try_cast(Amount_invested_monthly as float) Amount_invested_monthly
,Payment_Behaviour
,try_cast(Monthly_Balance as float) Monthly_Balance
from workspace.credit_risk.gold_credit_modeling_base
```
